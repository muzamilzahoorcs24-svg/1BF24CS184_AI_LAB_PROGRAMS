```python
class VacuumEnvironment:
    def __init__(self, location='A', status_a='Dirty', status_b='Dirty'):
        self.locations = {'A': status_a, 'B': status_b}
        self.agent_location = location

    def get_percept(self):
        """Returns current location and its status (e.g., ('A', 'Dirty'))."""
        return self.agent_location, self.locations[self.agent_location]

    def execute_action(self, action):
        """Executes the given action in the environment."""
        if action == 'Suck':
            self.locations[self.agent_location] = 'Clean'
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'

    def is_all_clean(self):
        return all(status == 'Clean' for status in self.locations.values())


class ReflexVacuumAgent:
    def choose_action(self, percept):
        """Reflex agent logic based on current percept."""
        location, status = percept

        if status == 'Dirty':
            return 'Suck'
        elif location == 'A':
            return 'Right'
        elif location == 'B':
            return 'Left'


# --- Simulation ---
def run_simulation(initial_location='A', status_a='Dirty', status_b='Dirty'):
    env = VacuumEnvironment(initial_location, status_a, status_b)
    agent = ReflexVacuumAgent()
    step = 1

    print(f"Initial State: Room A = {env.locations['A']}, Room B = {env.locations['B']}, Agent at = {env.agent_location}\n" + "-"*60)

    while not env.is_all_clean():
        percept = env.get_percept()
        action = agent.choose_action(percept)

        print(f"Step {step}:")
        print(f"  Percept: Location = {percept[0]}, Status = {percept[1]}")
        print(f"  Action: {action}")

        env.execute_action(action)

        print(f"  State After Action: Room A = {env.locations['A']}, Room B = {env.locations['B']}, Agent at = {env.agent_location}")
        print("-" * 60)
        step += 1

    print("Result: Both rooms are clean!")

run_simulation(initial_location='A', status_a='Dirty', status_b='Dirty')
```

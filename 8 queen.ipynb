```python
import random
import math

def calculate_cost(board):
    conflicts = 0
    n = len(board)
    for c1 in range(n):
        for c2 in range(c1 + 1, n):
            r1, r2 = board[c1], board[c2]
            if r1 == r2 or abs(r1 - r2) == abs(c1 - c2):
                conflicts += 1
    return conflicts

def get_neighbor(board):
    neighbor = list(board)
    col = random.randint(0, 7)
    current_row = neighbor[col]
    new_row = random.randint(0, 7)
    while new_row == current_row:
        new_row = random.randint(0, 7)
    neighbor[col] = new_row
    return neighbor

def simulated_annealing(initial_board=None, initial_temp=10.0, alpha=0.99, min_temp=0.0001, max_iterations=100000):
    current_board = list(initial_board) if initial_board is not None else [random.randint(0, 7) for _ in range(8)]
    current_cost = calculate_cost(current_board)
    T = initial_temp
    for step in range(max_iterations):
        if current_cost == 0 or T < min_temp:
            break
        neighbor = get_neighbor(current_board)
        neighbor_cost = calculate_cost(neighbor)
        delta_E = neighbor_cost - current_cost
        if delta_E < 0 or random.random() < math.exp(-delta_E / T):
            current_board = neighbor
            current_cost = neighbor_cost
        T *= alpha
    return current_board, current_cost, step

def print_board(board):
    print("\nChessboard Visualization:")
    for row in range(8):
        print(" ".join("Q" if board[col] == row else "." for col in range(8)))

if __name__ == "__main__":
    start_board = [0, 1, 2, 3, 4, 5, 6, 7]
    print(f"Initial Board: {start_board}")
    print(f"Initial Conflicts: {calculate_cost(start_board)}")
    solution, final_cost, iterations = simulated_annealing(initial_board=start_board)
    print("\n--- Results ---")
    print(f"Final Board: {solution}\nFinal Cost: {final_cost}\nTotal Iterations: {iterations}")
    if final_cost == 0:
        print("Success: Found a valid 8-Queens solution!")
        print_board(solution)
```
